**Ejercicio del slide:** *Un granjero tiene cincuenta animales entre conejos y gallinas. Si la cantidad de patas de los animales es ciento cuarenta, ¿cuántos conejos y cuántas gallinas tiene el granjero?*

Una posible solución (con $c$ conejos y $g$ gallinas: $c + g = 50$ y $4c + 2g = 140$):

````python
animales = 50
patas = 140

# 4c + 2g = patas  y  c + g = animales  ->  2c = patas - 2*animales
conejos = (patas - 2 * animales) // 2 # solo se guarda la parte entera de la división
gallinas = animales - conejos

print("Conejos:", conejos)
print("Gallinas:", gallinas)
````

### Operadores aritméticos

| Operador | Descripción |

## 4. Evaluación de secuencias de expresiones y traza de un programa (slide 05)

Una asignación primero **lee** los valores de las variables en memoria, **evalúa** la expresión de la derecha y luego **actualiza** la variable de la izquierda. Se puede pensar en instantes de tiempo $t$ y $t+1$.

**Ejemplo 1:** con $x = 3$, $y = 4$, $z = -2$ en el instante $t$, la asignación `x = y + z - 2` deja $x = 0$ en $t + 1$.

````python
x = 3
y = 4
z = -2

x = y + z - 2
print(x, y, z)  # 0 4 -2
````

**Ejemplo 2:** `x = x + 3 - 2 * y` con $x = 3$ y $y = 5$ → $x_{t+1} = x_t + 3 - 2 y_t = -4$.

````python
x = 3
y = 5

x = x + 3 - 2 * y
print(x, y)  # -4 5
````

### Traza de un programa

La traza es una tabla con una columna para el tiempo $t$ (inicialmente 0) y una columna por variable; `—` indica variable no inicializada.

**Ejemplo:** la traza de `x = x + 3 - 2 * y` con $x = 3$ y $y = 5$:

| t | x | y |
|---|---|---|
| 0 | 3 | 5 |
| 1 | −4 | 5 |

**Ejemplo:** secuencia de instrucciones con $k = 1$:

```python
i = k + 1
j = 2 * k
i = i * k * j
j = j * k - i
```

| t | k | i | j |
|---|---|---|---|
| 0 | 1 | — | — |
| 1 | 1 | 2 | — |
| 2 | 1 | 2 | 2 |
| 3 | 1 | 4 | 2 |
| 4 | 1 | 4 | −2 |

Se puede reproducir la traza imprimiendo el estado después de cada instrucción:

````python
k = 1
print("t=0", dict(k=k))

i = k + 1
print("t=1", dict(k=k, i=i))

j = 2 * k
print("t=2", dict(k=k, i=i, j=j))

i = i * k * j
print("t=3", dict(k=k, i=i, j=j))

j = j * k - i
print("t=4", dict(k=k, i=i, j=j))
````

#### Ley de Coulomb

$$F = \kappa \frac{Q_1 Q_2}{r^2}, \qquad \kappa = 9 \times 10^9\ \frac{Nm^2}{C^2}$$

$\text{ley\_coulomb}: \mathbb{R} \times \mathbb{R} \times \mathbb{R} \to \mathbb{R}$

````python
KAPPA: float = 9E+9  # constante global

def ley_coulomb(Q1, Q2, r):
    modulo = KAPPA * Q1 * Q2 / r ** 2
    return modulo

# carga1 = float(input("Carga 1: "))
# carga2 = float(input("Carga 2: "))
# distancia = float(input("Distancia entre cargas: "))
carga1 = 1.6e-6
carga2 = 0.4e-6
distancia = 0.3

print("El módulo de la fuerza es:", end=" ")
print(ley_coulomb(carga1, carga2, distancia))
````

### Valor absoluto de un número

$$\text{valor\_absoluto}(x) = \begin{cases} x & \text{si } x \ge 0 \\ -x & \text{en otro caso} \end{cases}$$

````python
def valor_absoluto(x):
    if x >= 0:
        valor = x
    else:
        valor = -x
    return valor


# x = float(input("Ingrese un número real x: "))
x = -7.5
valor_abs = valor_absoluto(x)
print("El valor absoluto de " + str(x), end=" ")
print("es " + str(valor_abs))
````

### El máximo entre dos números

#### El descuento del día

- Más de 5 productos del mismo tipo → 5 % de descuento.
- Más de 10 → 10 %.
- Más de 20 → 20 %.

$$\text{pago\_final}(n, precio) = \begin{cases} n \cdot precio & n \le 5 \\ n \cdot precio \cdot 0.95 & 5 < n \le 10 \\ n \cdot precio \cdot 0.90 & 10 < n \le 20 \\ n \cdot precio \cdot 0.80 & \text{en otro caso} \end{cases}$$

````python
def pago_final(n, precio):
    if n <= 5:
        valor = n * precio
    elif 5 < n <= 10:
        valor = n * precio * 0.95
    elif 10 < n <= 20:
        valor = n * precio * 0.90
    else:
        valor = n * precio * 0.80
    return valor
````

#### El mínimo número positivo de la máquina

Se calculan los términos de una progresión geométrica $x_0 = 1$, $x_{n+1} = x_n / 2$ mientras el nuevo término sea positivo.

````python
def min_maquina():
    Xo = 1.0
    Xi = Xo / 2.0
    while Xi > 0.0:
        Xo = Xi
        Xi = Xo / 2.0
    return Xo


print("El mínimo número positivo", end=" ")
print("en esta máquina es:", min_maquina())
````

### El ciclo hacer-mientras (*do-while*)

La condición se verifica al **final** del ciclo, por lo que el bloque se ejecuta al menos una vez. Python **no** tiene esta instrucción; se aproxima con `while`.

**Aproximación 1** (duplicando el bloque antes del ciclo; no recomendada porque duplica código):

```python
<bloque_prev>
```

````python
def min_maquina():
    Xi = 1.0  # Valor inicial
    Xo = Xi
    Xi = Xo / 2.0
    while(Xi > 0.0):
        Xo = Xi
        Xi = Xo / 2.0
    return Xo


print("El mínimo número positivo", end=" ")
print("en esta máquina es:", min_maquina())
````

**Aproximación 2: variable bandera** (garantiza que se entre al ciclo al menos una vez sin duplicar código):

```python
<bloque_prev>
<inicia>
bandera = True
while(bandera or <cond>):
    bandera = False
    <bloque>
    <actualiza>
<bloque_sigui>
```

````python
def min_maquina():
    Xi = 1.0       # Valor inicial
    bandera = True  # Permite que se ejecute el ciclo al menos una vez
    while(bandera or Xi > 0.0):
        bandera = False  # Ya se entró al ciclo; luego sólo se evalúa la condición
        Xo = Xi
        Xi = Xo / 2.0
    return Xo


print("El mínimo número positivo", end=" ")
print("en esta máquina es:", min_maquina())
````

### Interrupción y reinicio: `continue`

`continue` interrumpe la iteración actual y reinicia la siguiente (si la condición es verdadera).

### La colección `range`

Colección de enteros en un intervalo semiabierto $[C, F)$ con incremento $I$:

- `range(N)`: $0, 1, \dots, N - 1$
- `range(C, F)`: $C, C + 1, \dots, F - 1$
- `range(C, F, I)`: $C, C + I, C + 2I, \dots$ sin incluir $F$

Si no se da el inicio es 0 y si no se da el incremento es 1. $C$, $F$, $I$ deben ser enteros.

Ejemplos del slide:

````python
print(list(range(6)))               # [0, 1, 2, 3, 4, 5]
print(list(range(10, 21)))          # [10, 11, ..., 20]
print(list(range(-7, 8)))           # [-7, ..., 7]
print(list(range(2, 11, 2)))        # [2, 4, 6, 8, 10]
print(list(range(11, 20, 2)))       # [11, 13, 15, 17, 19]
print(list(range(5, -1, -1)))       # [5, 4, 3, 2, 1, 0]
print(list(range(-2)))              # []
````

### La suma de los primeros n números naturales

$$\sum_{i=1}^{n} i = 1 + 2 + 3 + \cdots + n$$

Las dos funciones son equivalentes: `i` es la variable **contadora** y `s` la variable **acumuladora**.

````python
def suma_while(n):
    s = 0
    i = 1
    while(i <= n):
        s += i
        i += 1
    return s


def suma(n):
    s = 0
    for i in range(1, n + 1):
        s += i
    return s


print(suma_while(6), suma(6))
````

````python
def suma(n):
    s = 0
    for i in range(1, n + 1):
        s += i
    return s


def main():
    # n = int(input("n? = "))
    n = 6  # por ejemplo el número de caras de un dado
    print("n? =", n)
    print("La suma de los primeros n números es:", end=" ")
    print(suma(n))


main()
````

````python
import math


def factorial(k):
    f = 1
    for j in range(2, k + 1):
        f *= j
    return f


# 8. exponencial
def exp_aprox(x, n):
    s = 0.0
    for i in range(n + 1):
        s += x ** i / factorial(i)
    return s


# 9. seno
def sin_aprox(x, n):
    s = 0.0
    for i in range(n + 1):
        s += (-1) ** i * x ** (2 * i + 1) / factorial(2 * i + 1)
    return s


# 10. coseno
def cos_aprox(x, n):
    s = 0.0
    for i in range(n + 1):
        s += (-1) ** i * x ** (2 * i) / factorial(2 * i)
    return s


# 11. logaritmo natural (serie del slide)
def ln_aprox(x, n):
    z = (x ** 2 - 1) / (x ** 2 + 1)
    s = 0.0
    for i in range(n + 1):
        s += z ** (2 * i + 1) / (2 * i + 1)
    return s


# 12. arco tangente
def arctan_aprox(x, n):
    s = 0.0
    for i in range(n + 1):
        s += (-1) ** i * x ** (2 * i + 1) / (2 * i + 1)
    return s


print(exp_aprox(1, 15), math.exp(1))
print(sin_aprox(math.pi / 6, 10), math.sin(math.pi / 6))
print(cos_aprox(math.pi / 3, 10), math.cos(math.pi / 3))
print(ln_aprox(2.0, 20), math.log(2.0))
print(arctan_aprox(0.5, 20), math.atan(0.5))
````

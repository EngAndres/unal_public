### Variables

Una **variable** es un espacio de memoria donde se almacena un dato. Se declara con un nombre (un identificador válido) y, opcionalmente, un tipo: `x: T`.

Buena práctica: nombrar la variable de acuerdo con el papel que desempeña, por ejemplo `velocidad`, `espacio`, `masa`, `aceleracion`, `exponente`, `termino1`, `valor_maximo`, `area_circulo`, `nombre_estudiante`, `last_name`.

#### Reales (`float`)

Se representan con 64 bits (IEEE 754), con rango aproximado de $2.2250738585072014 \times 10^{-308}$ a $1.7976931348623157 \times 10^{308}$ y una precisión científica de 15 dígitos.

#### Booleanos (`bool`)

Sólo hay dos literales: `True` y `False`. Internamente `True` ⇔ 1 y `False` ⇔ 0.

#### Caracteres y cadenas de caracteres (`str`)

Un **carácter** es el elemento mínimo de información; en Python se representa como una cadena de un solo carácter. El esquema de codificación **ASCII** usa 7 bits (128 caracteres) en su versión original y 8 bits (256 caracteres) en la extendida.

Una **cadena** es una secuencia de cero o más caracteres, delimitada por `'` o `"`, y es **inmutable**.

## 4. Evaluación de secuencias de expresiones y traza de un programa (slide 05)

Una asignación primero **lee** los valores de las variables en memoria, **evalúa** la expresión de la derecha y luego **actualiza** la variable de la izquierda. Se puede pensar en instantes de tiempo $t$ y $t+1$.

**Ejemplo 1:** con $x = 3$, $y = 4$, $z = -2$ en el instante $t$, la asignación `x = y + z - 2` deja $x = 0$ en $t + 1$.

**Ejemplo 2:** `x = x + 3 - 2 * y` con $x = 3$ y $y = 5$ → $x_{t+1} = x_t + 3 - 2 y_t = -4$.

### Traza de un programa

La traza es una tabla con una columna para el tiempo $t$ (inicialmente 0) y una columna por variable; `—` indica variable no inicializada.

**Ejemplo:** la traza de `x = x + 3 - 2 * y` con $x = 3$ y $y = 5$:

| t | x | y |
|---|---|---|
| 0 | 3 | 5 |
| 1 | −4 | 5 |

```python
i = k + 1
j = 2 * k
i = i * k * j
j = j * k - i
```

## 8. Ciclo `for` y `range` (slide 10)

El ciclo **para** (`for`) tiene dos propósitos: ser una forma compacta de escribir un `while` (como en C++ o Java) y **iterar sobre los elementos de una colección**. En Python sólo existe el segundo uso:

```python
<bloque_prev>
for <elemento> in <coleccion>:
    <bloque>
<bloque_sigui>
```

Se lee: *"para cada elemento en la colección realice las instrucciones del bloque"*.

En C++/Java el `for` compacto sería:

```cpp
int suma(int n){
    int s = 0;
    for(int i = 1; i <= n; i++){
        s = s + i;
    }
    return s;
}
```

### La colección `range`

Colección de enteros en un intervalo semiabierto $[C, F)$ con incremento $I$:

- `range(N)`: $0, 1, \dots, N - 1$
- `range(C, F)`: $C, C + 1, \dots, F - 1$
- `range(C, F, I)`: $C, C + I, C + 2I, \dots$ sin incluir $F$

Si no se da el inicio es 0 y si no se da el incremento es 1. $C$, $F$, $I$ deben ser enteros.
